## 🔪 특정 행과 열을 지정하여 선택하기 (loc, iloc)
더 구체적으로 원하는 위치의 데이터를 잘라내고 싶을 때 사용합니다. **loc**과 **iloc**의 차이를 이해하는 것이 중요합니다.

- df.loc[]: 이름(label) 기반으로 선택합니다. loc[행 이름, 열 이름] 형식으로 사용합니다.

- df.iloc[]: 순서(integer position) 기반으로 선택합니다. iloc[행 순서, 열 순서] 형식으로 사용하며, 컴퓨터가 숫자를 세는 방식처럼 0부터 시작합니다.

1. 행 선택하기

- 하나의 행 선택: df.loc[2] 또는 df.iloc[2]

- 여러 행 선택 (리스트 사용): df.loc[[2, 5, 10]] 또는 df.iloc[[2, 5, 10]]

- 연속된 범위의 행 선택 (슬라이싱):

   df.loc[2:5] → 2번부터 5번까지 (5를 포함)

   df.iloc[2:5] → 2번부터 4번까지 (5를 포함하지 않음)

2. 행과 열 함께 선택하기

[행, 열] 형식으로 쉼표(,)를 사용하여 구분합니다.

```python
# 예시 데이터 (위에서 계속)
# 이름(loc) 기반 선택
# 2번 행의 'col2' 열 선택
print(df.loc[    ,    ])

# 2번부터 5번 행까지, 'col1'과 'col2' 열 선택
print(df.loc[    ,    ])


# 순서(iloc) 기반 선택
# 3번째(순서 2) 행의 2번째(순서 1) 열 선택
print(df.iloc[    ,    ])

# 3번째부터 5번째 행까지, 1번째 열 선택
print(df.iloc[    ,    ])
```
