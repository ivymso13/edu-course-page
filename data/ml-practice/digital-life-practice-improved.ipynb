자기점검 함수는 답이 비어 있으면 안내를 표시하고, 입력한 결과가 문제와 맞는지 확인합니다.

```python
import pandas as pd
from IPython.display import display

df = None

def check_answer(value, expected, label):
    if value is None:
        print(f"{label}: 아직 코드를 작성하지 않았습니다.")
        return
    if isinstance(expected, (pd.DataFrame, pd.Series)):
        correct = isinstance(value, type(expected)) and value.equals(expected)
    else:
        try:
            correct = bool(value == expected)
        except (ValueError, TypeError):
            correct = False
    print(f"{label}: {'확인 완료!' if correct else '다시 확인하세요. 문제의 행·열·조건을 살펴보세요.'}")

def ready():
    if not isinstance(df, pd.DataFrame):
        print("먼저 1단계에서 digital_life.csv를 df에 불러오세요.")
        return False
    required = ["ID", "name", "gender", "phone", "sns", "stress"]
    if list(df.columns) != required or len(df) != 20:
        print("digital_life.csv 원본을 불러왔는지 확인하세요. 기본 실습 중 df를 다른 표로 바꾸지 않습니다.")
        return False
    return True
```
